# July 8, Tuesday plots (12SL Balloon)

```python
import numpy as np
import matplotlib.pyplot as plt
import gspread
from google.colab import auth
from google.auth import default

auth.authenticate_user()
creds, _ = default()
gc = gspread.authorize(creds)

worksheet = gc.open('Sample2').get_worksheet(0)
rows = worksheet.col_values(1)

# Convert to float safely
data = []
for val in rows:
    try:
        data.append(float(val))
    except:
        continue

data = np.array(data)

# Plot raw data
plt.figure(figsize=(14, 5))
plt.plot(data, linewidth=0.8, color='blue')
plt.title('Raw Arduino Serial Data')
plt.xlabel('Sample Index (every ~0.5s)')
plt.ylabel('Sensor Value')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Downsample for readability if needed
skip = 10
plt.figure(figsize=(14, 5))
plt.plot(data[::skip], linewidth=1, color='green')
plt.title(f'Arduino Data (every {skip}th point)')
plt.xlabel('Sample Index')
plt.ylabel('Sensor Value')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import gspread
from google.colab import auth
from google.auth import default

# Authenticate and connect
auth.authenticate_user()
creds, _ = default()
gc = gspread.authorize(creds)

worksheet = gc.open('Sample2').get_worksheet(0)
rows = worksheet.get_all_values()
rows = np.array(rows)

# Extract column index 3 and 4 (4th and 5th columns)
data_col4 = []
data_col5 = []

for row in rows:
    try:
        data_col4.append(float(row[3]))  # index 3 = 4th column
        data_col5.append(float(row[4]))  # index 4 = 5th column
    except:
        continue  # skip row if invalid

# Convert to numpy arrays
data_col4 = np.array(data_col4)
data_col5 = np.array(data_col5)

# Make sure both columns are the same length
min_len = min(len(data_col4), len(data_col5))
data_col4 = data_col4[:min_len]
data_col5 = data_col5[:min_len]
time = np.arange(min_len) * 0.5  # assuming 0.5s interval

# Plot both on the same figure
plt.figure(figsize=(14, 5))
plt.plot(time, data_col4, label='Column 4', color='blue')
plt.plot(time, data_col5, label='Column 5', color='orange')
plt.title('Sensor Data from Columns 4 and 5')
plt.xlabel('Time (seconds)')
plt.ylabel('Sensor Value')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Normalize both datasets to start at 1
data_col4_norm = (data_col4 - data_col4[0]) / (data_col4[0])
data_col5_norm = (data_col5 - data_col5[0]) / (data_col5[0])

plt.figure(figsize=(14, 5))
plt.plot(time, data_col4_norm, label='Col 4 (normalized)', color='blue')
plt.plot(time, data_col5_norm, label='Col 5 (normalized)', color='orange')
plt.title('Normalized Decay for Slope Comparison')
plt.xlabel('Time (seconds)')
plt.ylabel('Normalized Value (starts at 0)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
