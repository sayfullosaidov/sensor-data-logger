## 1. **ISA Model (International Standard Atmosphere)**

The **ISA model** is a mathematical model used by engineers and scientists to describe how **temperature**, **pressure**, and **density** of the atmosphere change with altitude.

https://eaglepubs.erau.edu/introductiontoaerospaceflightvehicles/chapter/international-standard-atmosphere-isa/

It's a simplification of Earth's real atmosphere, but quite accurate up to \~11 km altitude (Troposphere).

### In the troposphere (up to 11,000 m), two main assumptions:

* **Temperature drops linearly** with altitude:

  $$
  T(h) = T_0 - L \cdot h
  $$

  where:

  * $T_0 =$ ground temperature (usually 288.15 K or 293.15 K = 20°C)
  * $L = 0.0065 \, \text{K/m}$ (lapse rate = how fast temp drops with height)
  * $h =$ altitude in meters

* **Pressure drops exponentially**, derived using hydrostatics and ideal gas law:

  $$
  P(h) = P_0 \cdot \left(1 - \frac{L \cdot h}{T_0} \right)^{\frac{g \cdot M_{\text{air}}}{R \cdot L}}
  $$

  where:

  * $P_0 = 101325 \, \text{Pa}$ (sea level pressure)
  * $g = 9.80665 \, \text{m/s}^2$
  * $M_{\text{air}} = 0.0289644 \, \text{kg/mol}$
  * $R = 8.314462618 \, \text{J/mol·K}$

---

## 2. **Air and Helium Density at Altitude**

Now that we have **$P(h)$** and **$T(h)$**, we can compute **density** using the **ideal gas law**:

$$
\rho = \frac{P \cdot M}{R \cdot T}
$$

Apply this to:

* **Air**:

  $$
  \rho_{\text{air}}(h) = \frac{P(h) \cdot M_{\text{air}}}{R \cdot T(h)}
  $$
* **Helium**:

  $$
  \rho_{\text{He}}(h) = \frac{P(h) \cdot M_{\text{He}}}{R \cdot T(h)}
  $$

  where:

  * $M_{\text{He}} = 0.0040026 \, \text{kg/mol}$

**Observation**:
Even though both densities drop at altitude due to lower pressure, helium is **always lighter** than air. That’s why it **floats**.

---

## 3. **Buoyant Force at Altitude**

**Archimedes’ Principle** says:

> An object in a fluid is lifted by a force equal to the weight of the fluid it displaces.

For a balloon in air, the **buoyant force** is:

$$
F_b = \rho_{\text{air}}(h) \cdot V(h) \cdot g
$$

But the balloon is also **weighed down** by:

* The mass of the **helium inside**:

  $$
  W_{\text{He}} = \rho_{\text{He}}(h) \cdot V(h) \cdot g
  $$
* The **payload** and balloon material:

  $$
  W_{\text{payload}} = m_{\text{payload}} \cdot g
  $$

So, the **net lift** becomes:

$$
F_{\text{net}}(h) = F_b - W_{\text{He}} - W_{\text{payload}} = \left(\rho_{\text{air}} - \rho_{\text{He}} \right) \cdot V(h) \cdot g - m_{\text{payload}} \cdot g
$$

---

## 4. **Neutral Buoyancy Condition**

You want your balloon to **float at a certain altitude**, not rise further or fall. That means:

$$
F_{\text{net}}(h) = 0
\Rightarrow
\left( \rho_{\text{air}}(h) - \rho_{\text{He}}(h) \right) \cdot V(h) \cdot g = m_{\text{payload}} \cdot g
$$

Cancel out $g$ (same on both sides), and solve for helium volume at altitude:

$$
V_{\text{He}}(h) = \frac{m_{\text{payload}}}{\rho_{\text{air}}(h) - \rho_{\text{He}}(h)}
$$

---

## 5. **How to Convert This to Ground Volume**

You’re doing your experiment at **room conditions**, so you need to figure out:

**What volume of helium at room conditions (P₀, T₀) would expand to the required Vₕₑ at altitude?**

Use ideal gas law (PV = nRT):

$$
\text{Since } n_{\text{He}} = \frac{P \cdot V}{R \cdot T}, \text{ number of moles is conserved}
\Rightarrow
P_0 \cdot V_{\text{ground}} / T_0 = P(h) \cdot V(h) / T(h)
$$

Solve for ground volume:

$$
V_{\text{ground}} = V_{\text{He}}(h) \cdot \frac{P(h)}{P_0} \cdot \frac{T_0}{T(h)}
$$

This tells you **exactly how much helium to fill now**, so that **at high altitude**, your balloon expands to the volume required to **float**.


```python
import numpy as np

# Constants
g = 9.80665                # m/s²
R = 8.314462618            # J/(mol·K)
M_air = 0.0289644          # kg/mol
M_He = 0.0040026           # kg/mol
L = 0.0065                 # K/m (temperature lapse rate)
T0 = 293.15                # K (ground temperature: 20°C)
P0 = 101325                # Pa (ground pressure)

# Inputs
target_altitude = 300    # meters
payload_mass = 0.012 + 0.0061  # kg (payload + balloon)
P_ground = P0
T_ground = T0

# Step 1: Calculate temperature and pressure at altitude (ISA model)
T_alt = T0 - L * target_altitude
P_alt = P0 * (1 - (L * target_altitude) / T0) ** (g * M_air / (R * L))

# Step 2: Densities at altitude
rho_air = P_alt * M_air / (R * T_alt)
rho_He = P_alt * M_He / (R * T_alt)

# Step 3: Required helium volume at altitude for neutral buoyancy
delta_rho = rho_air - rho_He
if delta_rho <= 0:
    raise ValueError("No buoyancy possible: helium denser than air at this altitude.")

V_He_alt = payload_mass / delta_rho  # in m³

# Step 4: Convert to ground volume using Ideal Gas Law
V_He_ground = V_He_alt * (P_alt / P_ground) * (T_ground / T_alt)

# Output in liters
print(f"Target altitude: {target_altitude} m")
print(f"Air density at altitude: {rho_air:.4f} kg/m³")
print(f"Helium density at altitude: {rho_He:.4f} kg/m³")
print(f"Required helium volume at altitude: {V_He_alt*1000:.2f} L")
print(f"Required helium volume to fill at ground: {V_He_ground*1000:.2f} L")
```

## You can find the leakage rate

Since you're using a **load cell**, you're seeing the **apparent weight** of the helium balloon system, and over time, the **measured lift (or negative weight)** is decreasing due to helium escaping.

If your graph shows a **linear decrease in apparent lift**, then:

$$
\text{Leakage rate} = \frac{dW_{\text{net}}}{dt} \quad \text{(grams per second)}
$$

Since the **net upward force** is directly related to how much **helium is inside**, and that relates to **moles**, you can say:

$$
\frac{dW_{\text{net}}}{dt} \propto \frac{dn_{\text{He}}}{dt}
$$

More precisely, using buoyancy:

$$
F_{\text{net}}(t) = \left( \rho_{\text{air}} - \rho_{\text{He}}(t) \right) V_{\text{He}}(t) g - m_{\text{payload}} g
$$

And since:

$$
\rho_{\text{He}}(t) = \frac{P(t) M_{\text{He}}}{RT(t)}
\quad \text{and} \quad
V_{\text{He}}(t) = \frac{n_{\text{He}}(t) RT(t)}{P(t)}
$$

You get:

$$
\rho_{\text{He}}(t) V_{\text{He}}(t) = n_{\text{He}}(t) M_{\text{He}} = \text{mass of helium at time t}
$$

So the **loss of lift** is directly tied to the **loss of helium mass (or moles)**.

---

## 2. **Update the model to account for leakage**

Now let’s define:

* $n_{\text{He}}(0)$: initial number of moles of helium
* $r_{\text{leak}}$: leakage rate in **moles per second**
* $t$: time in seconds

Then:

$$
n_{\text{He}}(t) = n_{\text{He}}(0) - r_{\text{leak}} \cdot t
$$

And everything else (like $V_{\text{He}}(t), \rho_{\text{He}}(t), F_{\text{net}}(t)$) becomes **time-dependent**.

---

### Revised formulas with leakage

Let’s define:

* $P(h), T(h)$: pressure and temperature at altitude $h$
* $n_{\text{He}}(t) = n_0 - r_{\text{leak}} \cdot t$

Then:

1. **Volume of helium** at altitude $h$ and time $t$:

$$
V_{\text{He}}(h, t) = \frac{n_{\text{He}}(t) \cdot R \cdot T(h)}{P(h)}
$$

2. **Density of helium**:

$$
\rho_{\text{He}}(h, t) = \frac{P(h) \cdot M_{\text{He}}}{R \cdot T(h)}
\quad \text{(This actually doesn’t change with n)}
$$

But we can also use:

$$
\text{mass}_{\text{He}}(t) = n_{\text{He}}(t) \cdot M_{\text{He}}
$$

3. **Buoyant force**:

$$
F_b(h, t) = \rho_{\text{air}}(h) \cdot V_{\text{He}}(h, t) \cdot g
$$

4. **Weight of helium**:

$$
W_{\text{He}}(t) = n_{\text{He}}(t) \cdot M_{\text{He}} \cdot g
$$

5. **Net force**:

$$
F_{\text{net}}(h, t) = F_b(h, t) - W_{\text{He}}(t) - W_{\text{payload}}
$$

---

```python
import numpy as np

# Constants
g = 9.80665                # m/s²
R = 8.314462618            # J/(mol·K)
M_air = 0.0289644          # kg/mol
M_He = 0.0040026           # kg/mol
L = 0.0065                 # K/m
T0 = 293.15                # K (20°C)
P0 = 101325                # Pa

# Inputs
target_altitude = 300                 # meters
payload_mass = 0.012 + 0.0061         # kg (payload + balloon)
leakage_rate_g_per_s = 0.0001         # grams per second (example)
leakage_duration_sec = 3 * 3600       # 3 hours

# Step 1: Altitude temperature and pressure (ISA model)
T_alt = T0 - L * target_altitude
P_alt = P0 * (1 - (L * target_altitude) / T0) ** (g * M_air / (R * L))

# Step 2: Densities at altitude
rho_air = P_alt * M_air / (R * T_alt)
rho_He = P_alt * M_He / (R * T_alt)
delta_rho = rho_air - rho_He
if delta_rho <= 0:
    raise ValueError("No buoyancy possible: helium denser than air at this altitude.")

# Step 3: Volume of helium needed at altitude for neutral buoyancy
V_He_alt = payload_mass / delta_rho  # m³

# Step 4: Helium moles needed at altitude
n_He_final = (P_alt * V_He_alt) / (R * T_alt)

# Step 5: Add leakage correction
leakage_rate_mol_per_s = (leakage_rate_g_per_s / 1000) / M_He
n_He_initial = n_He_final + leakage_rate_mol_per_s * leakage_duration_sec

# Step 6: Compute required ground volume to contain n_He_initial
V_He_ground = (n_He_initial * R * T0) / P0  # m³

# Final clean output
print(f"Target altitude: {target_altitude} m")
print(f"Air density at altitude: {rho_air:.4f} kg/m³")
print(f"Helium density at altitude: {rho_He:.4f} kg/m³")
print(f"Required helium volume at altitude: {V_He_alt * 1000:.2f} L")
print(f"Required helium volume to fill at ground (after {leakage_duration_sec/3600:.1f} h of leakage): {V_He_ground * 1000:.2f} L")
```
